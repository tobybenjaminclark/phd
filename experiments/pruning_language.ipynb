```python
from __future__ import annotations
from dataclasses import dataclass
from enum import Enum
from typing import Protocol



# Operators
class CmpOp(Enum):
    GT = ">"
    LT = "<"
    EQ = "="
    GE = "≥"
    LE = "≤"


class ArithOp(Enum):
    ADD = "+"
    SUB = "-"
    MUL = "×"
    DIV = "÷"



# Boolean Expression
class BooleanExpr(Protocol):
    pass

@dataclass(frozen=True)
class And(BooleanExpr):
    left: BooleanExpr
    right: BooleanExpr
    def __str__(self): return f"{self.left} ∧ {self.right}"


@dataclass(frozen=True)
class Or(BooleanExpr):
    left: BooleanExpr
    right: BooleanExpr
    def __str__(self): return f"{self.left} ∨ {self.right}"

@dataclass(frozen=True)
class Not(BooleanExpr):
    inner: BooleanExpr
    def __str__(self): return f"¬ ({self.inner})"

@dataclass(frozen=True)
class Cmp(BooleanExpr):
    left: ArithExpr
    op: CmpOp
    right: ArithExpr
    def __str__(self): return f"{self.left} {self.op.value} {self.right}"



# Arithmetic Expression
class ArithExpr(Protocol):
    pass

@dataclass(frozen=True)
class Binary(ArithExpr):
    left: ArithExpr
    op: ArithOp
    right: ArithExpr
    def __str__(self): return f"{self.left} {self.op.value} {self.right}"

@dataclass(frozen=True)
class Number(ArithExpr):
    value: int
    def __str__(self): return f"{self.value}"

@dataclass(frozen=True)
class Symbol(ArithExpr):
    name: str
    def __str__(self): return f"{self.name}"



rule = Not(
    Cmp(
        left=Binary(
            left=Symbol("x"),
            op=ArithOp.ADD,
            right=Number(3)
        ),
        op=CmpOp.GT,
        right=Binary(
            left=Symbol("y"),
            op=ArithOp.MUL,
            right=Number(2)
        )
    )
)

print(rule)
```
